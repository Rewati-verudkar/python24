# Functions

```python
##1.Upper case

ab="Live the life you love"
```

```python
print(ab.upper())
```

```python
##Lower case
ab="Live the life you love"
print(ab.lower())
```

```python
##strip:- The strip() method removes any whitespace from the beginning or the end.
ab="  Live the life you love  "
print(ab.strip())
```

```python
##Replace:- The replace() method replaces a string with another string:
ab="Live the life you love"
print(ab.replace("e","y"))
```

```python
##split:- The split() method splits the string into substrings if it finds instances of the separator:
ab="Live the life you love"
rv=ab.split()
print(rv)

rt=ab.split(",")
print(rt)
```

```python
a = "Have a good day!"
b = a.split(",")
print(b)
```

# String Formatting

```python
## F - formating
name = "Rewati"
age = 25
txt = f"My name is {name}, I am {age}."
print(txt)
```

### Placeholder and modifier

```python
price = 59
amt = f"The price is {price} dollars"
print(amt)

#with decimal:--

price = 59
amt = f"The price is {price:.2f} dollars"
print(amt)
```

### Math operation

```python
math = f"The price is {30 * 9} Rupees"
print(math)
add = f"The price is {30 + 30} Rupees"
print(add)
div = f"The price is {30 / 9} Rupees"
print(div)
sub = f"The price is {30 -9} Rupees"
print(sub)
```
